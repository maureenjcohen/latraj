## Verification Testing on Balloon Kernel

### (a) Terminal Velocity

"Hold the balloon limp and check that the steady ascent rate matches the analytic terminal velocity. Cross-check against the paper: 320 g of free lift at rho = 0.85 kg/m^3 gives w_term = 0.69 m/s, and Fig. 12 shows the real prototype climbing at roughly 1-1.5 m/s. Same order of magnitude; do not expect better, since our masses are assumed."

```python
%run run_venus_simulation.py
```

When the balloon arrests at around rho = 0.85 kg/m^3, the terminal velocity stays roughly around 0.69 m/s. The terminal velocity varies due to the rising and falling motion of the balloon around maximum inflation.

### (b) Ceiling Behaviour

"Confirm the balloon rises monotonically while limp and arrests only when V reaches V_infl, at the altitude where rho_atm * V_infl = m_total. A balloon that settles below full inflation is a bug -- see "The physics you should expect"."


```python
%run run_venus_simulation.py
```

```python
from heatmap import double_heatmap
ds = xr.open_zarr('outputs/KernelTest.zarr')
double_heatmap(ds)
```

The particles appear to rise monotonically, with some spread in the trajectories. Once V = V_infl, the particles arrest at around 55km, where the estimated mass (rho_atm * V_infl) is roughly equal to m_total (62kg).

### (c) Timestep Convergence

"Halve dt_inner and confirm the trajectory changes negligibly. If it does not, the sub-stepping is insufficient."

### (c2) JIT vs. Scipy Agreement

"Run the same short case with pclass=BalloonParticle (JIT) and with a ScipyParticle-based equivalent, and confirm thetrajectories match. This catches silent mistranslation in the inner loop, which is the failure mode JIT is most likely to produce here. Do this once, early, on a short run -- not repeatedly."


```python
from heatmap import double_heatmap
ds = xr.open_zarr('outputs/KernelTest.zarr')
print('JITParticle 2-day run')
double_heatmap(ds)
```

```python
from heatmap import double_heatmap
ds = xr.open_zarr('outputs/KernelTest.zarr')
print('ScipyParticle 2-day run')
double_heatmap(ds)
```

The trajectories appear largely identical for JITParticle and ScipyParticle on a 2-day run.

### (d) Compute Tau

"Compute tau_horiz for the actual configuration and confirm it is far below the outer dt, which is the assumption justifying passive horizontal advection. If it is not, that decision needs revisiting."

### (e) Zero-buoyancy Degenerate Case

"Set the net lift to exactly zero and confirm the balloon tracks W_atm, i.e. reduces to the existing passive tracer. This is the cleanest single test that the kernel does not double-count the vertical wind."

```python
%run run_venus_simulation.py
```

When the net lift is set to zero, the value of w_rel_old is small, sometimes reaching a magnitude of 1.e-05. Therefore, the difference between W_atm and W_bal is negligible, and the balloon tracks W_atm.
