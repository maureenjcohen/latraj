## Vega Balloon Comparisons

```python
import pandas as pd # column 12 and 14
import xarray as xr
import matplotlib.pyplot as plt
import math
df = pd.read_csv('data/vg1bl_rdr.dat', sep="\s+", usecols=[11, 13], names=['Lat', 'Lon'], header=None)
xvega, yvega = df['Lon']*(180/math.pi), df['Lat']*(180/math.pi)

ds = xr.open_zarr('outputs/Vega_run.zarr')
x, y = ds.lon.values, ds.lat.values

plt.scatter(xvega, yvega, label='Vega 1 Balloon')
plt.scatter(x, y, label='Balloon Particle')
plt.xlabel('Longitude (degrees)')
plt.ylabel('Latitude (degrees)')
plt.title('Vega Balloon 1 vs Balloon Particle Trajectory')
plt.legend()
plt.show()
```

```python
import pandas as pd # column 12 and 14
import xarray as xr
import matplotlib.pyplot as plt
import math
df = pd.read_csv('data/vg1bl_rdr.dat', sep="\s+", usecols=[11, 13], names=['Lat', 'Lon'], header=None)
xvega, yvega = df['Lon']*(180/math.pi), df['Lat']*(180/math.pi)

ds = xr.open_zarr('outputs/Vega2_run.zarr')
x, y = ds.lon.values, ds.lat.values

plt.scatter(xvega, yvega, label='Vega 2 Balloon', linewidths=1)
plt.scatter(x, y, label='Balloon Particle', linewidths=1)
plt.xlabel('Longitude (degrees)')
plt.ylabel('Latitude (degrees)')
plt.title('Vega Balloon 2 vs Balloon Particle Trajectory')
plt.legend()
plt.show()
```
